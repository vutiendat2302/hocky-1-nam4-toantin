## 5. Vẽ hình tròn ở chính giữa ảnh, chỉ có viền

```python
# use library circle
img_bgr = img_bgr_original.copy()
size = 200

height, width = img_bgr.shape[:2]
center_x = width // 2
center_y = height // 2

radius = 250 
cv2.circle(img_bgr, (center_x, center_y), radius, (255,0,0), 10)

img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
# use algorithm: (x - a)^2 + (y-b)^2 = r^2 
img_bgr = img_bgr_original.copy()
size = 200

height, width = img_bgr.shape[:2]
center_x = width // 2
center_y = height // 2

radius = 250 
thickness = 10 

for y in range (height): 
    for x in range (width):

        # Khoảng cách từ pixel (x, y) đến tâm đường tròn (center_x, center_y)
        distance = ((x-center_x)**2 + (y - center_y)**2)**0.5 

        if radius - thickness <= distance <= radius:
            img_bgr[y, x] = (0, 0, 255)
            
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

## 6. Thay đổi giá trị điểm ảnh để có chữ T hoa màu đỏ ở giữa ảnh.
Ví dụ

image.png

```python
img = img_bgr_original.copy()
h, w, _ = img.shape

# Kích thước chữ T
thickness = 100
top_width = 200
height = 300

# Tọa độ trung tâm ảnh
cx = w // 2
cy = h // 2

# Thanh ngang phía trên của chữ T
x1 = cx - top_width // 2
x2 = cx + top_width // 2
y1 = cy - height // 2
y2 = y1 + thickness

img[y1:y2, x1:x2] = (0, 0, 255)

# Thanh dọc của chữ T
x1 = cx - thickness // 2
x2 = cx + thickness // 2
y1 = cy - height // 2
y2 = cy + height // 2

img[y1:y2, x1:x2] = (0, 0, 255)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

- Sử dụng điểm ảnh

```python
img = img_bgr_original.copy()

H, W, C = img.shape

xc = W // 2
yc = H // 2

for y in range(H):
    for x in range(W):

        # Thanh ngang của chữ T
        ngang = (
            xc - 100 <= x < xc + 100
            and
            yc - 100 <= y < yc - 70
        )

        # Thanh dọc của chữ T
        doc = (
            xc - 15 <= x < xc + 15
            and
            yc - 100 <= y < yc + 100
        )

        if ngang or doc:
            # OpenCV lưu theo BGR
            img[y, x] = [0, 0, 255]

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

## 7. Cho 2 ảnh I1 và I2 cùng kích thước. Tính ảnh I_kq như sau:

*Với mọi tọa độ x, y: I_kq(x, y) = ( I1(x, y) + I2(x, y) )/2*

Dự đoán nội dung hiển thị trên ảnh I_kq và lập trình kiểm chứng.

Result: Ảnh sẽ là sự pha trộn của hai ảnh 


```python
I1 = cv2.imread("bt1.png")
I2 = cv2.imread("i2.png")

# OpenCV đọc ảnh theo BGR → đổi sang RGB để hiển thị đúng màu
I1 = cv2.cvtColor(I1, cv2.COLOR_BGR2RGB)
I2 = cv2.cvtColor(I2, cv2.COLOR_BGR2RGB)

print("Kích thước I1:", I1.shape)
print("Kích thước I2:", I2.shape)
```

```python
H, W, C = I1.shape

Ikq = np.zeros_like(I1)

for y in range(H):
    for x in range(W):
        for c in range(C):
            Ikq[y, x, c] = (
                int(I1[y, x, c]) + int(I2[y, x, c])
            ) / 2
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(I1)
plt.title("I1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(I2)
plt.title("I2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(Ikq)
plt.title("Ikq = (I1 + I2) / 2")
plt.axis("off")

plt.show()
```
