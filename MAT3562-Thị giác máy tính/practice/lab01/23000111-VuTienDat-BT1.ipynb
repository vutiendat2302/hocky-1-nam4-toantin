```python
import cv2 
import matplotlib.pyplot as plt
```

```python
  # 1. Đọc file ảnh                                                                                                                                                            
image_path = 'anh_mau.jpg'  # Đường dẫn ảnh của bạn                                                                                                                          
img_bgr = cv2.imread(image_path)  
```

```python
# Lưu ý: OpenCV đọc theo thứ tự màu BGR, cần chuyển sang RGB để Matplotlib vẽ đúng màu                                                                                       
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

```python
# 2. Hiển thị ảnh trong JupyterLab                                                                                                                                           
plt.imshow(img_rgb)                                                                                                                                                          
plt.title("Ảnh hiển thị bằng Matplotlib")
plt.axis('on')  # Hoặc 'off' nếu không muốn hiện trục toạ độ
plt.show()

# 3. Kích thước ảnh: height (chiều cao), width (chiều rộng), channels (số kênh màu)
height, width, channels = img_bgr.shape
print(f"Chiều cao (Height): {height} px")
print(f"Chiều rộng (Width): {width} px")
print(f"Số kênh màu (Channels): {channels}")

# 4. Giá trị của một điểm ảnh tại toạ độ (x=50, y=100)
# Lưu ý: ma trận ảnh lập chỉ mục theo cú pháp [dòng, cột] = [y, x]
x, y = 50, 100
pixel_bgr = img_bgr[y, x]  # [Blue, Green, Red]
print(f"Giá trị điểm ảnh tại toạ độ (x={x}, y={y}): {pixel_bgr}")
```
