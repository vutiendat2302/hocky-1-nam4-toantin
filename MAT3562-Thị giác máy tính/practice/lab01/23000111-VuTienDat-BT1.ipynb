## 1. Cài Python và OpenCV hoặc dùng GG Colab 

```python
import cv2 
import matplotlib.pyplot as plt
```

## 2. Đọc file ảnh và hiển thị được ảnh 

```python
# Read file                                                                                                                                                            
image_path = 'bt1.png'                                                                                                                         
img_bgr_original = cv2.imread(image_path)  
img_bgr = img_bgr_original.copy()
```

```python
# Color Picture 
plt.imshow(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
# Grayscale 
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap = "gray")
plt.axis("off")
plt.show()
```

```python
# Binary Image 
_, img_binary = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(img_binary, cmap="gray")
plt.axis("off")
plt.show()
```

## 3. Sau khi hiển thị được ảnh ra, in chiều dài, chiều rộng, chiều cao, in giá trị của một điểm ảnh. 

```python
height, width, channels = img_bgr.shape

print("width:", width)
print("height:", height)
print("channels:", channels)

x, y = 100, 50
print("Value pixel of (x, y):", img_bgr[y, x])
```

```python
print(img_bgr.shape)
print("Chiều rộng:", width)
print("Chiều cao:", height)
print("Số kênh màu:", channels)
```

## 4. Trên ảnh đã mở được rồi, tô màu ở góc trên bên trái (góc 1/4) thành màu xanh lá cây 

```python
img_bgr = img_bgr_original.copy()
height, width = img_bgr.shape[:2]
img_bgr[0:height//2, 0:width//2] = (0,255,0)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

## 5. Vẽ hình vuông ở chính giữa ảnh, chỉ có viền

```python
img_bgr = img_bgr_original.copy()
height, width = img_bgr.shape[:2]

size = 500

x1 = (width - size) // 2
y1 = (height - size) // 2

x2 = x1 + size
y2 = y1 + size

cv2.rectangle(
    img_bgr,
    (x1, y1),
    (x2, y2),
    (0, 0, 255),
    10
)

img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
img_bgr = img_bgr_original.copy()
size = 200

x1 = (width - size) // 2
y1 = (height - size) // 2

x2 = x1 + size 
y2 = y1 + size 

img_bgr[y1:y1+10, x1:x2] = (0,255, 0)
img_bgr[y2-10:y2, x1:x2] = (0, 255, 0)
img_bgr[y1:y2, x1:x1+10] = (0, 255, 0)
img_bgr[y1:y2, x2-10:x2] = (0, 255, 0)

img_rgb_square = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb_square)
plt.axis("off")
plt.show()
```

## 5. Vẽ hình tròn ở chính giữa ảnh, chỉ có viền

```python
# use library circle
img_bgr = img_bgr_original.copy()
size = 200

height, width = img_bgr.shape[:2]
center_x = width // 2
center_y = height // 2

radius = 250 
cv2.circle(img_bgr, (center_x, center_y), radius, (255,0,0), 10)

img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
# use algorithm: (x - a)^2 + (y-b)^2 = r^2 
img_bgr = img_bgr_original.copy()
size = 200

height, width = img_bgr.shape[:2]
center_x = width // 2
center_y = height // 2

radius = 250 
thickness = 10 

for y in range (height): 
    for x in range (width):

        # Khoảng cách từ pixel (x, y) đến tâm đường tròn (c
        distance = ((x-center_x)**2 + (y - center_y)**2)**0.5 

        if radius - thickness <= distance <= radius:
            img_bgr[y, x] = (0, 0, 255)
            
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```
